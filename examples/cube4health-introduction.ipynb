The cube4health is a data acquisition, processing, and publishing toolkit designed to streamline the integration of health, climate, and drone imagery data from Brazil and any other country adhering to standardized protocols [**ToDo:** add link for protocols page documentation!]. It enables the efficient handling of heterogeneous data sources into a unified structure, supporting scalable workflows for data ingestion, transformation, and dissemination.

This Jupyter Notebook gives an overview of how to use the cube4health to process and publish data products from the Earth Observation Data Cubes tuned for Health Response Systems (EODCtHRS), a <a href="https://harmonize-tools.org" target="_blank">HARMONIZE project</a> component.

## Installation
<hr style="border:1px solid #274ad4;">

The cube4health relies primarily on the Geospatial Data Abstraction Library (GDAL) for raster processing. Please follow the instructions below to install the package.

### Install GDAL library and its development header files on your system (Debian based distributions of Linux):

```
sudo apt-get update && sudo apt-get upgrade
sudo apt-get install -y g++ && sudo apt-get install -y gdal-bin libgdal-dev
```

### Install Python and PIP:

The package manager (PIP) will be used on the command line to download and install Python packages and their requisite dependencies.


```
sudo apt-get install python3
sudo apt-get install python3-pip
```


### Install the Python dependencies and build numpy-based raster support:

```
export GDAL_VERSION=$(gdal-config --version)
export CPLUS_INCLUDE_PATH=/usr/include/gdal
export C_INCLUDE_PATH=/usr/include/gdal
pip3 install --upgrade "pip<=25.2" wheel numpy
pip3 install --use-pep517 --no-build-isolation --no-cache-dir --force-reinstall gdal[numpy]==`gdal-config --version`
```

Verify that numpy-based raster support has been installed:

```
python -c "from osgeo import gdal, gdal_array; print(gdal.__version__)"
```

**Obs.**: If this command raises an ImportError, numpy-based raster support has not been properly installed. Please see these [related issues and solutions](https://github.com/Harmonize-Brazil/cube4health/blob/main/docs/ISSUES.rst)!



### Install cube4health from Github:

```
pip3 install git+https://github.com/Harmonize-Brazil/cube4health.git
```

## Usage
<hr style="border:1px solid #274ad4;">


The package ```cube4health``` has an entry point based on the Click package for the execution as a command-line tool using a Terminal that allows processing and publishing data produced in the context of the EODCtHRS. It is important to note that the publishing process is based on predefined protocols [ToDo: add link for protocols page documentation] and additional services, such as BDC-STAC, Geoserver, and Titiler. Please see the documentation available on the [Cube4Health GitHub page](https://harmonize-brazil.github.io/cube4health/index.html).

If you want to know the cube4health version, use the option --version:



We have specific module (subpackage) for processing and publishing each type of data  (``eddpr`` - drone, ``ehipr`` - health, and ``eclimpr`` - climate). For example, try eddpr to see parameter requirements:




This is a complete example of using the eddpr module to process and publish data at localhost:

```
cube4health run --module eddpr  --server_type localhost 
--root_path /home/user/Desktop/HARMONIZE-Br_Project/src/FieldWorkCampaigns 
--data_path_output /home/user/Docker-Compose/geoserver/data  --publish_data True
```

The ``eddpr`` module requires the use of protocol[**ToDo** add link for drone protocol] that shows how to provide images and auxiliary information from fieldwork campaigns available in the folder specified through the --root_path option. The creation of image or mosaic collections depends on [template files](https://github.com/Harmonize-Brazil/cube4health/tree/main/src/cube4health/eddpr/templates) in JSON format for each device (e.g., Phantom 3 Advanced, Mavic 3M, Mavic 3 Enterprise), type (scene, mosaic, or thermal), and flight height.
